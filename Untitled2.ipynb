```python
#etapa 1 - msg de boas vindas
print("====== bem vindo ao sacolão da Generation ======")

# etapa 2 - carregar dados - criar lista
produtos = ["Banana","Melancia", "Maçã"]
precos = [5,19,12.90]

# etapa 3 - estrutura condicionada
# etapa 3.1 - vamos mostrar os produtos
for i in produtos:
  print(i)

# for de duas listas
for i,j in zip(produtos, precos):
  print(f"o produto {i} custa R${j}")

resposta = input('deseja cadastrar novo produto?')

if resposta.lower() == "sim":
  novo_produto = input("informe o nome do produto")
  novo_preco = float(input("informe o valor"))
  print("produto cadastrado com sucesso")

  # sempre que quisermos colocar um novo valor na lista
  #devemos usar o método de anexo
  #lista.anexo ( novo_valor)
  produtos.append(novo_produto)
  precos.append(novo_preco)
  for i,j in zip(produtos, precos):
    print(f"o produto {i} custa R$ {j}")
```
